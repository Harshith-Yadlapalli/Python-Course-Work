```python
for i in range(3):
    for j in range(3):
        print("*", end=" ")
    print()
```

```python
n=5
for i in range(n):
    for j in range(n):
        if i==j or (i+j)==n-1:
            print("*", end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
n=5
for i in range(n):
    for j in range(n):
        if i==(n//2) or j==(n//2):
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
        
```

```python
for i in range(4):
    for j in range(4):
        if (i+j)%2==0:
            print("1",end=" ")
        else:
            print("0",end=" ")
    print()
```

```python
for i in range(4):
    for j in range(4):
        if (i-j)%2==0:
            print("1",end=" ")
        else:
            print("0",end=" ")
    print()
```
