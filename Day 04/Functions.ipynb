```python
# Functions 
# with return with parammeters
def addition(x,y):
    return x+y
addition(20,30)
```

```python
# with return withour parameters
def mul():
    a=8
    b=9
    return a*b
mul()
```

```python
# Without return with parameter
def sub(x,y):
    print(x-y)
sub(90,30)
    
```

```python
# Without return without parameters
def div():
    a=45
    b=5
    print(a/b)
div()
```

```python
#Positional arguments
def area(a,b):
    return a*b
area(10,20)
```

```python
#keyword arguments
def add(a,b):
    print(a,b)
    return a+b
c=98
d=67
#add(d=a,c=b)
add(b=c,a=d)
```

```python
#default arguments
def default(a,b=0,c=5):
    return a+b*c
default(10,20,30)
```

```python
# Variable length arguments
def addition(*args):
    return sum(args)
print(addition(10, 20))
print(addition(10, 20, 30, 40))
```

```python
# Keyword variable length arguments
def details(**kwargs):
    for key, value in kwargs.items():
        print(key, value)
details(name="Harsh", age=22, city="Hyderabad")
```
